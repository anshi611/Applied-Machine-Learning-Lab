```python

import matplotlib.pyplot as plt  # For creating static, animated, and interactive visualizations
import numpy as np  # For high-performance numerical array operations
import pandas as pd  # For data manipulation and analysis structures (DataFrames/Series)
import seaborn as sns  # For statistical data visualization built on top of Matplotlib
```

```python
# Set a random seed so numbers are reproducible every time we run this
np.random.seed(42)

```

```python
# np.array(): Creates a multi-dimensional array from a list
data_list = [10, 20, 30, 40, 50]
np_array = np.array(data_list)

```

```python
np_array
```

```python
# np.arange(): Generates an array with evenly spaced values within a given interval [start, stop, step]
time_steps = np.arange(0, 10, 1)  # array([0, 1, 2, ..., 9])

# np.linspace(): Generates a specified number of evenly spaced values over a specified range [start, stop, num]
x_vals = np.linspace(0, 5, 100)  # 100 points between 0 and 5

```

```python

# Mathematical and statistical functions on arrays
array_mean = np.mean(x_vals)  # np.mean(): Computes the arithmetic mean (average)
array_std = np.std(x_vals)  # np.std(): Computes the standard deviation
array_sum = np.sum(np_array)  # np.sum(): Computes the sum of all elements
```

```python
# Create a sample dictionary dataset
raw_data = {
    "Category": ["A", "B", "A", "C", "B", "C", "A", "B"],
    "Values": np.random.randint(50, 100, 8),  # Random integers between 50 and 100
    "Scores": np.random.uniform(1.0, 5.0, 8),  # Random floats between 1.0 and 5.0
}

# pd.DataFrame(): Converts a dictionary or array into a 2D tabular data structure with labeled axes
df = pd.DataFrame(raw_data)

# df.head(): Returns the first n rows of the DataFrame (default 5)
print("First 3 rows of DataFrame:\n", df.head(3))

# df.info(): Provides a concise summary of the DataFrame, including data types and non-null counts
print("\nDataFrame Info:")
df.info()

# df.describe(): Generates descriptive statistics (mean, min, max, percentiles) for numerical columns
print("\nDescriptive Statistics:\n", df.describe())

# df.groupby(): Groups data using a mapper or by a series of columns to perform aggregations
grouped_df = df.groupby("Category")["Values"].mean().reset_index()
# .mean(): Computes mean for groups, .reset_index(): Converts grouped result back to a standard DataFrame
print("\nGrouped Mean Values:\n", grouped_df)
```

```python
# plt.figure(): Creates a new figure window to control size and resolution
plt.figure(figsize=(8, 4))

# plt.plot(): Draws a line plot of y versus x
plt.plot(
    x_vals, np.sin(x_vals), label="Sine Wave", color="blue", linestyle="--"
)  # np.sin() calculates sine values

# plt.title(), plt.xlabel(), plt.ylabel(): Adds text labels and titles to the axes
plt.title("Matplotlib Line Plot Example")
plt.xlabel("X Axis Values")
plt.ylabel("Sine Value")

# plt.legend(): Places a legend on the plot to identify curves/elements
plt.legend()

# plt.grid(): Adds grid lines to the plot area for easier reading
plt.grid(True)

# plt.show(): Displays the constructed plot window
plt.show()
```

```python
# Set a visual theme for Seaborn aesthetics
sns.set_theme(style="whitegrid")

# plt.figure(): Setup canvas size for the Seaborn plot
plt.figure(figsize=(8, 4))

# sns.boxplot(): Draws a box-and-whisker plot to show distributions across categories
sns.boxplot(x="Category", y="Values", data=df, palette="Set2")

# Add title and labels via Matplotlib interface (Seaborn builds on top of Matplotlib)
plt.title("Seaborn Box Plot Example: Values by Category")
plt.xlabel("Category")
plt.ylabel("Values")

# Display the Seaborn plot
plt.show()
```
